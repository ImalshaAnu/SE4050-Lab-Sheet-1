```python
# Pandas for managing datasets
import pandas as pd
```

```python
# Matplotlib for additional customization
from matplotlib import pyplot as plt
%matplotlib inline
# Seaborn for plotting and styling
import seaborn as sns
```

```python
!wget -q https://elitedatascience.com/wp-content/uploads/2022/07/Pokemon.csv
# Read dataset
df = pd.read_csv('Pokemon.csv', index_col=0, encoding='latin')
```

```python

# Display first 5 observations
df.head()
```

Scatter Plot

```python

# Scatterplot arguments
sns.lmplot(x='Attack', y='Defense', data=df,
           fit_reg=True, # No regression line
           hue='Stage')   # Color by evolution stage
```

Box Plots

```python
# Boxplot
plt.figure(figsize=(9,6)) # Set plot dimensions
sns.boxplot(data=df)
```

```python

# Preprocess DataFrame
stats_df = df.drop(['Total', 'Stage', 'Legendary'], axis=1)

# New boxplot using stats_df
plt.figure(figsize=(9,6)) # Set plot dimensions
sns.boxplot(data=stats_df)
```

Violin Plot

```python
# Violin plot
plt.figure(figsize=(12,8)) # Set plot dimensions
sns.violinplot(x='Type 1', y='Attack', data=df)
```

Heat Map

```python
# Select only numeric columns
numeric_df = stats_df.select_dtypes(include=['number'])

# Calculate correlations
corr = numeric_df.corr()

corr
```

```python
# Heatmap
plt.figure(figsize=(9,8))
sns.heatmap(corr)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Load the Pokemon dataset
!wget -q https://elitedatascience.com/wp-content/uploads/2022/07/Pokemon.csv
```

```python
#load the dataset
df = pd.read_csv('Pokemon.csv', index_col=0, encoding='latin')
```

```python
df.head()
```

```python
#Select only required features
stats_df = df[['HP',
               'Attack',
               'Defense',
               'Sp. Atk',
               'Sp. Def',
               'Speed']]
```

```python
#Display selected data
stats_df.head()
```

```python
#Pearson Correlation
#Calculate Pearson correlation
pearson_corr = stats_df.corr(method='pearson')

pearson_corr
```

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    pearson_corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Pearson Correlation Heatmap")

plt.show()
```

```python
#Spearman Rank Correlation
#Calculate Spearman correlation
spearman_corr = stats_df.corr(method='spearman')

spearman_corr
```

```python
#Create Spearman Heatmap
plt.figure(figsize=(8,6))

sns.heatmap(
    spearman_corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Spearman Rank Correlation Heatmap")

plt.show()
```
